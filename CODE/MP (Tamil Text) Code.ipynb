```python
import unicodedata
import string
import nltk
from nltk.tokenize import word_tokenize
from bs4 import BeautifulSoup
import re
from collections import Counter

nltk.download('punkt')

def clean_text(text):
    """Preprocess and clean the input Tamil text."""
    text = unicodedata.normalize("NFKC", text)
    text = BeautifulSoup(text, "html.parser").get_text()  
    text = re.sub(r"http\S+|www\S+|https\S+", "", text, flags=re.MULTILINE)  
    text = text.translate(str.maketrans("", "", string.punctuation))  
    text = text.lower()  
    return text

tamil_stopwords = {"மற்றும்", "இது", "என்று", "என", "ஒரு", "ஆகிய", "உள்ள", "நான்", "நீ", "அவன்", "அவர்", "இங்கே"}

def tokenize_and_remove_stopwords(text):
    """Tokenize text and remove Tamil stopwords."""
    words = word_tokenize(text)  
    words = [word for word in words if word not in tamil_stopwords]  
    return words

def get_word_frequency(words):
    """Calculate word frequency in the text."""
    return Counter(words)

def preprocess_tamil_text(text):
    """Full preprocessing pipeline for Tamil text."""
    cleaned_text = clean_text(text)  
    tokens = tokenize_and_remove_stopwords(cleaned_text)  
    word_freq = get_word_frequency(tokens) 
    return tokens, word_freq

tamil_text = """தமிழ் இலக்கியத்தின் வளர்ச்சி மற்றும் அதன் முக்கியத்துவம்

தமிழ் மொழி, உலகின் பழமையான மொழிகளில் ஒன்றாக இருக்கிறது. அதன் இலக்கிய வரலாறு சங்க காலத்திலிருந்து தொடங்கி இன்றுவரை தொடர்கிறது. தமிழ் இலக்கியம் பல்வேறு பரிணாமங்களை கடந்து வளர்ந்துள்ளது. சங்க இலக்கியம் தமிழர்களின் வாழ்க்கை முறையை, பண்பாட்டை, கலை மற்றும் மரபுகளை விவரிக்கிறது. அந்தக் காலத்தில் எழுத்து முறைகள் மிகவும் நடைமுறைக்கு உரியவையாக இருந்தன. சங்க இலக்கியத்திற்குப் பிறகு, முத்தமிழின் வளர்ச்சி தொடங்கியது.

சித்தர் பாடல்கள், தேவாரம், திருவாசகம் போன்ற நூல்கள் தமிழ் இலக்கியத்தின் ஆன்மீக வளர்ச்சிக்குப் பெரிதும் உதவின. இவை தமிழ் மக்களின் ஆன்மீக விழிப்புணர்வை தூண்டியது. பழந்தமிழ் இலக்கியம் மிகுந்த பொறுமை, நேர்மை, உழைப்பு ஆகியனவற்றின் மதிப்பை வலியுறுத்தியது.

மத்தியகால தமிழ் இலக்கியத்தில், கம்பராமாயணம், சிலப்பதிகாரம், மணிமேகலை போன்ற நூல்கள் தமிழின் செழுமையை வெளிப்படுத்துகின்றன. குறிப்பாக, சிலப்பதிகாரம் தமிழ்ச் சமூகத்தின் நீதியை, பெண்களின் முக்கியத்துவத்தை, காதலின் தூய்மையை விவரிக்கின்றது. கம்பராமாயணம் ராமாயணக் கதையை தமிழ் மொழியில் அழகாக எடுத்துச் சொல்லும் ஒரு சிறப்பு நூலாகும்.

நவீன தமிழ் இலக்கியம் 19ஆம் நூற்றாண்டில் முக்கியமான மாற்றங்களை கண்டது. சுப்பிரமணிய பாரதி, பாரதிதாசன் போன்ற கவிஞர்கள் தமிழுக்கு புதிய உயிர்கொடுத்தனர். பாரதி தனது கவிதைகளில் பெண்களின் விடுதலை, சமத்துவம், தேசப்பற்று போன்ற விசயங்களை எடுத்துக்காட்டினார். பாரதிதாசன் சமத்துவம், சாதி ஒழிப்பு போன்ற பல சமூக சிக்கல்களை தனது கவிதைகளில் விவரித்தார்.

இன்றைய தமிழ் இலக்கியம் பல்வேறு துறைகளில் வளர்ச்சியடைந்துள்ளது. தமிழ் நாவல்கள், சிறுகதைகள், கவிதைகள், நாடகங்கள், கட்டுரைகள் போன்றவை புதிய கருத்துக்களை முன்வைக்கின்றன. தமிழ் திரைப்படங்களும் இலக்கிய வளர்ச்சிக்கு பெரிதும் உதவுகின்றன. பல திரைப்படங்கள் இலக்கிய ஆக்கங்களை அடிப்படையாக கொண்டு உருவாக்கப்படுகின்றன.

தமிழ் மொழியின் வளர்ச்சிக்கு உலகெங்கும் உள்ள தமிழ் ஆர்வலர்கள் பல முயற்சிகளை மேற்கொள்கிறார்கள். பல்வேறு தமிழ் அமைப்புகள், தமிழ் இணைய இதழ்கள், தமிழ் ஆராய்ச்சிகள் மூலம் தமிழின் பெருமை பரவியுள்ளது. இதன் மூலம், தமிழ் மொழியும் அதன் இலக்கியமும் எதிர்காலத்திலும் வளர்ச்சியடையும் என்று நம்பலாம்."""

tokens, word_freq = preprocess_tamil_text(tamil_text)

print("\nTokens:")
print(tokens)

print("\nWord Frequency:")
print(word_freq)
```
