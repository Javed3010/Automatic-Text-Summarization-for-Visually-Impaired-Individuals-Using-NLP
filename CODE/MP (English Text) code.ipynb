```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
```

```python
tokens = [token.text.lower() for token in doc 
          if not token.is_stop and 
          not token.is_punct and 
          token.text !='\n']
```

```python
tokens
```

```python
tokens1=[]
stopwords = list(STOP_WORDS)
allowed_pos = ['ADJ','PROPN','VERB','NOUN']
for token in doc:
    if token.text in stopwords or token.text in punctuation:
        continue
    if token.pos_ in allowed_pos:
        tokens1.append(token.text)
```

```python
tokens1
```

```python
from collections import Counter
```

```python
word_freq = Counter(tokens)
```

```python
word_freq
```

```python
max_freq = max(word_freq.values())
```

```python
max_freq
```

```python
for word in word_freq.keys():
    word_freq[word] = word_freq[word]/max_freq
```

```python
word_freq
```

```python
sent_token = [sent.text for sent in doc.sents]
```

```python
sent_token
```

```python
sent_score = {}
for sent in sent_token:
    for word in sent.split():
        if word.lower() in word_freq.keys():
            if sent not in sent_score.keys():
                sent_score[sent] = word_freq[word]
            else:
                sent_score[sent] +=word_freq[word]
        print(word)
```

```python
sent_score
```

```python
import pandas as pd
```

```python
pd.DataFrame(list(sent_score.items()),columns=['Sentence','Score'])
```

                                                       EXTRACTIVE SUMMARIZATION 

```python
from heapq import nlargest
```

```python
num_sentences = 30
n = nlargest(num_sentences,sent_score,key=sent_score.get)
```

```python
m = " ".join(n) # OUTPUT
```

```python
display(m)
```

```python
len(m)
```

                                                         TEXT TO VOICE OUTPUT 

```python
from gtts import gTTS
from deep_translator import GoogleTranslator
import os

summary_text = m  

hindi_text = GoogleTranslator(source='en', target='hi').translate(summary_text)
tamil_text = GoogleTranslator(source='en', target='ta').translate(summary_text)

def text_to_speech(text, lang, filename):
    tts = gTTS(text=text, lang=lang)
    tts.save(filename)  
    os.system(f"start {filename}")  

user_choice = input("Please enter your preferred Language for Voice Output (English/Hindi/Tamil): ").strip()

if user_choice.lower() == "english":
    text_to_speech(summary_text, "en", "english.mp3")
elif user_choice.lower() == "hindi":
    text_to_speech(hindi_text, "hi", "hindi.mp3")
elif user_choice.lower() == "tamil":
    text_to_speech(tamil_text, "ta", "tamil.mp3")
else:
    print("The language chosen is not available. Sorry!")
```

```python
from rouge_score import rouge_scorer

def evaluate_summary(reference, generated):

    if isinstance(reference, list):
        reference = " ".join(str(item) for item in reference)  
    if isinstance(generated, list):
        generated = " ".join(str(item) for item in generated) 
    
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL', 'rougeLsum'], use_stemmer=True)
    scores = scorer.score(reference, generated)
    
    print("\nROUGE Scores:")
    print(f"ROUGE-1: Precision: {scores['rouge1'].precision:.4f}, Recall: {scores['rouge1'].recall:.4f}, F1: {scores['rouge1'].fmeasure:.4f}")
    print(f"ROUGE-2: Precision: {scores['rouge2'].precision:.4f}, Recall: {scores['rouge2'].recall:.4f}, F1: {scores['rouge2'].fmeasure:.4f}")
    print(f"ROUGE-L: Precision: {scores['rougeL'].precision:.4f}, Recall: {scores['rougeL'].recall:.4f}, F1: {scores['rougeL'].fmeasure:.4f}")
    print(f"ROUGE-LSUM: Precision: {scores['rougeLsum'].precision:.4f}, Recall: {scores['rougeLsum'].recall:.4f}, F1: {scores['rougeLsum'].fmeasure:.4f}")
    
    
    return scores

reference_summary = text
generated_summary = summary

rouge_results = evaluate_summary(reference_summary, generated_summary)
```
